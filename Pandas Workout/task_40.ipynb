```python
import numpy as np
import pandas as pd
import time
```

```python
df = pd.read_csv(
    'data/nyc_taxi_2019-07.csv',
    usecols=['tpep_pickup_datetime', 'passenger_count', 'trip_distance', 'total_amount'],
    parse_dates=['tpep_pickup_datetime']
)

df
```

```python
df['tpep_pickup_datetime'] = df['tpep_pickup_datetime'].dt.strftime('%d/%m/%y %Hh:%Mm:%Ss')
```
