```python
s = pd.Series(data=g.integers(40, 61, 10), index=months, dtype=np.int8)
s
```

```python
s.mean()
```

```python
avg = s.mean()
std = s.std()
```

```python
print(f'{avg} {std}')
```

```python
def get_grade_calc(avg, std):
    def get_grade(score):
        if score < avg - std:
            return 'D'
        elif avg - std <= score < avg:
            return 'C'
        elif avg <= score < avg + std:
            return 'B'
        else:
            return 'A'
    
    return get_grade 
```

```python
s_new = s.apply(get_grade_calc(avg, std))
s_new
```
