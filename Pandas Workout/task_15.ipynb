```python
payments = df.groupby('payment_type')
```

```python
cnt = payments['passenger_count'].count()
```

```python
cnt_share = cnt / payments['passenger_count'].count().sum() * 100
```

```python
amt = payments['total_amount'].sum()
```

```python
amt_share = amt / payments['total_amount'].sum().sum() * 100
```

```python
cnt_share.name = 'passenger_count_share'
```

```python
amt.name
```

```python
amt_share.name = 'total_amount_share'
```

```python
df1 = pd.concat(objs=[cnt, cnt_share, amt, amt_share], axis=1)
df1.loc[[1, 2], ['passenger_count_share', 'total_amount_share']].round(1)
```
