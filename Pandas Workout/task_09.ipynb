```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame(
    data=[
        {'product_id': 23, 'name': 'computer',       'wholesale_price': 500, 'retail_price': 1000, 'sales': 100},
        {'product_id': 96, 'name': 'Python Workout', 'wholesale_price': 35,  'retail_price': 75,   'sales': 1000},
        {'product_id': 97, 'name': 'Pandas Workout', 'wholesale_price': 35,  'retail_price': 75,   'sales': 500},
        {'product_id': 15, 'name': 'banana',         'wholesale_price': 0.5, 'retail_price': 1,    'sales': 200},
        {'product_id': 87, 'name': 'sandwich',       'wholesale_price': 3.0, 'retail_price': 5,    'sales': 300},
    ]
)
df
```

```python
df['current_net'] = (df['retail_price'] - df['wholesale_price']) * df['sales']
df
```

```python
df['after_15'] = (df['retail_price'] - df['wholesale_price']) * df['sales'] * 0.85
```

```python
df['after_20'] = (df['retail_price'] - df['wholesale_price']) * df['sales'] * 0.8
```

```python
df['after_25'] = (df['retail_price'] - df['wholesale_price']) * df['sales'] * 0.75
```

```python
df
```

```python
df[['current_net', 'after_15', 'after_20', 'after_25']].sum()
```

```python
def get_alter_25(profit):
    if profit > 20000:
        return profit * 0.75
    else:
        return profit
```

```python
df['alter_after_25'] = df['current_net'].apply(get_alter_25)
df
```

```python
df[['current_net', 'after_15', 'after_20', 'after_25', 'alter_after_25']].sum()
```

```python
def get_tax_rate(retail_price):
    if retail_price > 80:
        return 0.25
    elif 30 <= retail_price <= 80:
        return 0.1
    else:
        return 0.0
```

```python
df['new_tax'] = (1 - df['retail_price'].apply(get_tax_rate)) * df['current_net']
df
```

```python
df[['current_net', 'after_15', 'after_20', 'after_25', 'alter_after_25', 'new_tax']].sum()
```
