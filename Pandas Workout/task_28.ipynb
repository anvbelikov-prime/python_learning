```python
import string

def normalize_name(name):
    if not isinstance(name, str):
        return name

    new_name = ''

    for l in name.strip().upper():
        if l in string.ascii_uppercase:
            new_name += l

    return new_name
```

```python
df['Vehicle Make'].apply(normalize_name).nunique()
```

```python
def normalize_name_v2(name):
    if not isinstance(name, str):
        return name

    new_name = []
    
    for l in name.strip().upper():
        if l.isalpha():
            new_name.append(l)

    return ''.join(new_name)
```

```python
df['Vehicle Make'].apply(normalize_name_v2).nunique()
```
