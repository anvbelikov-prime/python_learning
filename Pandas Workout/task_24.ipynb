```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('data/olympic_athlete_events.csv', usecols=['Age', 'Height', 'Team', 'Year', 'Season', 'Sport', 'Medal'])
df
```

```python
df = df.query('Year >= 1980').copy()
```

```python
df = df.query('Team.isin(["Great Britain", "France", "United States", "Russia", "China", "India"])').copy()
```

```python
df['Team'].unique()
```

```python
sorted(df['Year'].unique())
```

```python
df.pivot_table(index='Team', columns='Year', values='Age', aggfunc='mean')
```

```python
df.pivot_table(index='Year', columns='Team', values='Age', aggfunc='mean')
```

```python
df.pivot_table(index='Sport', columns='Year', values='Height', aggfunc='max', fill_value=0)
```

```python
df.query('Medal.notnull()').pivot_table(index='Year', columns='Team', values='Medal', aggfunc='count')
```

```python
df.pivot_table(index='Year', columns='Team', values='Medal', aggfunc='count')
```
