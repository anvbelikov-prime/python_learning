```python
import numpy as np
import pandas as pd
```

```python
df1 = pd.read_csv('data/nyc_taxi_2019-07.csv', usecols=['passenger_count', 'total_amount', 'payment_type'])
df1['year'] = 2019
df1
```

```python
df1.count()
```

```python
df2 = pd.read_csv('data/nyc_taxi_2020-07.csv', usecols=['passenger_count', 'total_amount', 'payment_type'])
df2['year'] = 2020
df2
```

```python
df = pd.concat(objs=[df1, df2], ignore_index=True)
df
```

```python
df['year'].value_counts()
```

```python
(df['year'].value_counts().loc[2020] - df['year'].value_counts().loc[2019]) / df['year'].value_counts().loc[2019] * 100
```

```python
years = df.groupby('year')
```

```python
tay = years['total_amount'].sum()
```

```python
(tay.loc[2020] - tay.loc[2019]) / tay.loc[2019] * 100
```
