```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv(
    'data/sat-scores.csv',
    usecols=[
        'Year',
        'State.Code',
        'Total.Math',
        'Family Income.Less than 20k.Math',
        'Family Income.Between 20-40k.Math',
        'Family Income.Between 40-60k.Math',
        'Family Income.Between 60-80k.Math',
        'Family Income.Between 80-100k.Math',
        'Family Income.More than 100k.Math'
    ]
)
df
```

```python
df = df.rename(columns={
        'Family Income.Less than 20k.Math': 'income<20k',
        'Family Income.Between 20-40k.Math': '20k<income<40k',
        'Family Income.Between 40-60k.Math': '40k<income<60k',
        'Family Income.Between 60-80k.Math': '60k<income<80k',
        'Family Income.Between 80-100k.Math': '80k<income<100k',
        'Family Income.More than 100k.Math': 'income>100k'
    }
)
```

```python
df
```

```python
df.groupby('Year')[['income<20k', '20k<income<40k', '40k<income<60k', '60k<income<80k', '80k<income<100k', 'income>100k']].mean().sort_index()
```

```python
df.groupby('Year')[['income<20k', '20k<income<40k', '40k<income<60k', '60k<income<80k', '80k<income<100k', 'income>100k']].mean().transpose()
```

```python
df.groupby('Year')[['income<20k', '20k<income<40k', '40k<income<60k', '60k<income<80k', '80k<income<100k', 'income>100k']].mean().transpose().pct_change()
```

```python
df.groupby('Year')[['income<20k', '20k<income<40k', '40k<income<60k', '60k<income<80k', '80k<income<100k', 'income>100k']].mean().transpose().pct_change().transpose().mean()
```

```python
df.groupby('Year')[['income<20k', '20k<income<40k', '40k<income<60k', '60k<income<80k', '80k<income<100k', 'income>100k']].mean().transpose().pct_change().mean(axis=1)
```
