```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv(
    'data/nyc-parking-violations-2020.csv',
    low_memory=False
)

df.head()
```

```python
df.info()
```

```python
print(f'{df.memory_usage().sum():,}')
```

```python
cols = [
    'Plate ID',
    'Registration State',
    'Vehicle Make',
    'Vehicle Color',
    'Vehicle Body Type',
    'Violation Time',
    'Street Name',
    'Violation Legal Code',
]
```

```python
for col in cols:
    df[col] = df[col].astype('category')
```

```python
df.info()
```

```python
other_objects = list(df.select_dtypes(include=['object']).columns)
```

```python
other_objects
```

```python
other_objects.remove('Issue Date')
other_objects.remove('Time First Observed')
```

```python
other_objects
```

```python
for col in other_objects:
    df[col] = df[col].astype('category')
```

```python
df.info()
```

```python
print(f'{df.memory_usage().sum():,}')
```
