```python
import numpy as np
import pandas as pd
```

```python
import calendar
```

```python
idx = [calendar.day_name[i] for i in range(0, 7)] * 4
idx
```

```python
len(idx)
```

```python
g = np.random.default_rng(0)
```

```python
s = pd.Series(data=g.normal(20, 5, 28).round().astype(np.int8), index=idx)
s
```

```python
s.loc['Monday'].mean()
```

```python
s.loc[['Saturday', 'Sunday']].mean()
```
