```python
import numpy as np
import pandas as pd
```

```python
s = pd.read_csv('taxi-distance.csv', header=None).squeeze()
s
```

```python
def get_distance_category(distance):
    if distance <= 2:
        return 'short'
    elif 2 < distance <= 10:
        return 'medium'
    else:
        return 'long'
```

```python
s.apply(get_distance_category).value_counts()
```

```python
pd.cut(s, bins=[s.min(), 2, 10, s.max()], labels=['short', 'medium', 'long'], include_lowest=True).value_counts()
```
