```python
df['age'] = df['age'].fillna(df['age'].mean())
```

```python
df['home.dest'] = df['home.dest'].fillna(df['home.dest'].mode())
```

```python
df['home.dest'].mode()
```

```python
df = df.dropna(subset=['fare', 'embarked'])
```

```python
# 27.1
```

```python
idx = df['embarked'].unique()
```

```python
dest = []
```

```python
for e in idx:
    dest.append(df.loc[df['embarked'] == e, 'home.dest'].mode()[0])
```

```python
dest
```

```python
most_common_destinations = pd.Series(data=dest, index=idx)
most_common_destinations
```

```python
def get_dest(dest, embarked):
    if pd.isna(dest):
        return most_common_destinations.loc[embarked]
    else:
        return dest
```

```python
df.apply(lambda row: get_dest(row['home.dest'], row['embarked']), axis=1)
```
