```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_json('data/cities.json')
df
```

```python
df.dtypes
```

```python
df.groupby(by=['state'])['city'].count().sort_values().plot.bar(figsize=(15, 10))
```

```python
df.loc[df['growth_from_2000_to_2013'] == '', 'growth_from_2000_to_2013'] = '0.0'
```

```python
df['growth_from_2000_to_2013'] = df['growth_from_2000_to_2013'].str.replace('%', '')
```

```python
df['growth_from_2000_to_2013'] = df['growth_from_2000_to_2013'].astype(float)
```

```python
df.dtypes
```

```python
df.loc[df['state'] == 'Pennsylvania', ['city', 'growth_from_2000_to_2013']].sort_values(by=['growth_from_2000_to_2013']).plot.bar(x='city', figsize=(10,10))
```

```python
df.loc[df['state'] == 'Massachusetts', ['city', 'population']].set_index('city').plot.pie(y='population', legend=None, figsize=(10,10))
```

```python
df.plot.scatter(x='longitude', y='latitude', figsize=(15, 10))
```
