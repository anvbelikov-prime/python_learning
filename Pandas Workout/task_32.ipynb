```python
dataframes = []
for file in [file_name for file_name in os.listdir('data') if ('+' in file_name) or (',' in file_name)]:
    full_name, extension = file.split('.')
    city, state = full_name.split(',')
    city = city.replace('+', ' ')
    df_tmp = pd.read_csv(
        'data/' + file,
        usecols=[0, 1, 2],
        names=['date_time', 'max_temp', 'min_temp'],
        header=0
    )
    df_tmp['city'] = city.title()
    df_tmp['state'] = state.upper()
    dataframes.append(df_tmp)
```

```python
df = pd.concat(objs=dataframes, ignore_index=True)
df
```

```python
df.groupby(['state', 'city'])['date_time'].agg(['min', 'max'])
```

```python
df.groupby(['state', 'city'])['date_time'].agg(['min', 'max']).nunique()
```
