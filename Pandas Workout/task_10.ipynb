```python
import numpy as np
import pandas as pd
```

```python
df1 = pd.DataFrame(
    data=[
        {'product_id': 23, 'name': 'computer',       'wholesale_price': 500, 'retail_price': 1000, 'sales': 100},
        {'product_id': 96, 'name': 'Python Workout', 'wholesale_price': 35,  'retail_price': 75,   'sales': 1000},
        {'product_id': 97, 'name': 'Pandas Workout', 'wholesale_price': 35,  'retail_price': 75,   'sales': 500},
        {'product_id': 15, 'name': 'banana',         'wholesale_price': 0.5, 'retail_price': 1,    'sales': 200},
        {'product_id': 87, 'name': 'sandwich',       'wholesale_price': 3.0, 'retail_price': 5,    'sales': 300},
    ]
)
df1
```

```python
df2 = pd.DataFrame(
    data=[
        {'product_id': 24, 'name': 'phone', 'wholesale_price': 200, 'retail_price': 500},
        {'product_id': 16, 'name': 'apple', 'wholesale_price': 0.5, 'retail_price': 1},
        {'product_id': 17, 'name': 'pear',  'wholesale_price': 0.6, 'retail_price': 1.2},
    ]
)
df2
```

```python
df = pd.concat([df1, df2], ignore_index=True)
df
```

```python
df.loc[df['sales'].isnull(), 'sales'] = [100, 200, 75]
```

```python
df
```

```python
((df['retail_price'] - df['wholesale_price']) * df['sales']).sum()
```

```python
def get_department(name):
    if name in ('computer', 'phone'):
        return 'electronics'
    elif 'Workout' in name:
        return 'books'
    else:
        return 'food'
```

```python
df['department'] = df['name'].apply(get_department)
df
```

```python
df['current_net'] = (df['retail_price'] - df['wholesale_price']) * df['sales']
df
```

```python
dep = df.groupby('department')
```

```python
dep['current_net'].sum()
```
