```python
import numpy as np
import pandas as pd
```

```python
oecd_df = pd.read_csv(
    'data/oecd_locations.csv',
    header=None,
    index_col=0,
    names=['LOCATION', 'Country']
)

oecd_df
```

```python
oecd_tourism_df = pd.read_csv(
    'data/oecd_tourism.csv',
    usecols=['LOCATION', 'TIME', 'SUBJECT', 'Value'],
    index_col=['LOCATION']
)

oecd_tourism_df
```

```python
oecd_tourism_df = oecd_tourism_df[oecd_tourism_df['SUBJECT'] == 'INT-EXP']
oecd_tourism_df
```

```python
oecd_tourism_df = oecd_tourism_df.drop(['SUBJECT'], axis=1)
oecd_tourism_df
```

```python
len(oecd_tourism_df.groupby('LOCATION')['Value'].mean())
```

```python
tourism_spending = oecd_df.join(oecd_tourism_df).groupby('Country')['Value'].mean()
```

```python
tourism_spending
```

```python
wine_df = pd.read_csv(
    'data/winemag-150k-reviews.csv',
    usecols=['country', 'points']
)

wine_df
```

```python
wine_points = wine_df.groupby(by=['country'])['points'].mean().sort_values(ascending=False).to_frame()
```

```python
wine_points
```

```python
wine_points.join(tourism_spending, how='outer').corr()
```

```python
oecd_df = pd.read_csv(
    'data/oecd_locations.csv',
    header=None,
    names=['abbrev', 'country']
)

oecd_df
```

```python
oecd_tourism_df = pd.read_csv(
    'data/oecd_tourism.csv',
    usecols=['LOCATION', 'TIME', 'SUBJECT', 'Value']
).loc[lambda df_: df_['SUBJECT'] == 'INT-EXP'].drop('SUBJECT', axis=1)

oecd_tourism_df
```

```python
wine_df = pd.read_csv(
    'data/winemag-150k-reviews.csv',
    usecols=['country', 'points']
)

wine_df
```

```python
tourism_spendings = oecd_df.merge(oecd_tourism_df, left_on='abbrev', right_on='LOCATION').groupby('country')['Value'].mean().to_frame()
```

```python
tourism_spendings
```

```python
tourism_spendings.merge(wine_df.groupby('country')['points'].mean(), left_on='country', right_on='country', how='outer')
```
