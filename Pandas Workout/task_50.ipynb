```python
import numpy as np
import pandas as pd
```

```python
import timeit
```

```python
df = pd.read_csv(
    'data/nyc-parking-violations-2020.csv',
    usecols=[
        'Plate ID',
        'Registration State',
        'Plate Type',
        'Feet From Curb',
        'Vehicle Make',
        'Vehicle Color'
    ],
    low_memory=False
)

df
```

```python
df.columns
```

```python
df.columns = ['pid', 'state', 'ptype', 'make', 'color', 'feet']
```

```python
df
```

```python
%%timeit

len(df) - df.count()
```

```python
len(df) - df.count()
```

```python
%%timeit
df.set_index(keys=['state']).sort_index().loc[['NY', 'NJ', 'CT']]
```

```python
%%timeit
df.loc[df['state'].isin(['NY', 'NJ', 'CT'])]
```

```python
%%timeit
df.query('state.isin(["NY", "NJ", "CT"])')
```

```python
%%timeit
df.loc[(df['state'] == 'NY') | (df['state'] == 'NJ') | (df['state'] == 'CT')]
```

```python
%%timeit
df.query('(state == "NY") | (state == "NJ") | (state == "CT")')
```

```python
%timeit df.loc[df['state'] == 'NY']
%timeit df.query('state == "NY"')
%timeit df.loc[df.eval('state == "NY"')]
```

```python
%timeit df.loc[(df['state'] == 'NY') & (df['ptype'] == 'PAS')]
%timeit df.query('state == "NY" and ptype == "PAS"')
%timeit df.loc[df.eval('state == "NY" and ptype == "PAS"')]
```

```python
%timeit df.loc[(df['state'] == 'NY') & (df['ptype'] == 'PAS') & (df['color'] == 'WHITE')]
%timeit df.query('state == "NY" and ptype == "PAS" and color == "WHITE"')
%timeit df.loc[df.eval('state == "NY" and ptype == "PAS" and color == "WHITE"')]
```

```python
%timeit df.loc[(df['state'] == 'NY') & (df['ptype'] == 'PAS') & (df['color'] == 'WHITE') & (df['feet'] > 1)]
%timeit df.query('(state == "NY") and (ptype == "PAS") and (color == "WHITE") and (feet > 1)')
%timeit df.loc[df.eval('(state == "NY") and (ptype == "PAS") and (color == "WHITE") and (feet > 1)')]
```

```python
%timeit df.loc[(df['state'] == 'NY') & (df['ptype'] == 'PAS') & (df['color'] == 'WHITE') & (df['feet'] > 1) & (df['make'] == 'TOYOTA')]
%timeit df.query('(state == "NY") and (ptype == "PAS") and (color == "WHITE") and (feet > 1) and (make == "TOYOTA")')
%timeit df.loc[df.eval('(state == "NY") and (ptype == "PAS") and (color == "WHITE") and (feet > 1) and (make == "TOYOTA")')]
```
