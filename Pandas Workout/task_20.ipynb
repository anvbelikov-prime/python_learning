```python
df.loc[df['growth_from_2000_to_2013'] == '', 'growth_from_2000_to_2013'] = '0.0%'
```

```python
df['growth_from_2000_to_2013'] = df['growth_from_2000_to_2013'].str.replace('%', '').astype(np.float32) / 100
```

```python
df['growth_from_2000_to_2013'].describe()[['mean', '50%']]
```

```python
df['category'] = pd.cut(df['growth_from_2000_to_2013'], bins=[df['growth_from_2000_to_2013'].min(), 0, df['growth_from_2000_to_2013'].max()], labels=['neg', 'pos'], include_lowest=True)
```

```python
df['category'].value_counts()
```
