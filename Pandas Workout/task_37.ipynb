```python
words_france = df.loc[df['country'] == 'France']['description'].str.split().explode().str.strip().str.strip(string.punctuation).str.lower()
words_france.value_counts().head(10)
```

```python
def get_top_words(descs, n=10):
    words = descs.str.split().explode().str.strip().str.strip(string.punctuation).str.lower()
    words = words.loc[(words.str.len() >= 5) & ~(words.isin(['flavors', 'aromas', 'finish', 'palate', 'drink']))]
    return words.value_counts().head(n)
```

```python
get_top_words(df['description'])
```

```python
get_top_words(df.loc[df['country'] == 'France']['description'])
```

```python
get_top_words(df.loc[df['province'] != 'California']['description'])
```

```python
get_top_words(df.loc[df['variety'].isin(['Chardonnay', 'Sauvignon Blanc', 'Riesling'])]['description'])
```

```python
get_top_words(df.loc[df['variety'].isin(['Cabernet Sauvignon', 'Syrah', 'Merlot', 'Zinfandel'])]['description'])
```

```python
get_top_words(df.loc[df['variety'].isin(['Rosé'])]['description'])
```

```python
df['variety'].value_counts().head(5).index
```

```python
get_top_words(df.loc[df['variety'].isin(df['variety'].value_counts().head(5).index)]['description'])
```
