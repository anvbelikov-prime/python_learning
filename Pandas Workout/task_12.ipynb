```python
import numpy as np
import pandas as pd
```

```python
s1 = pd.read_csv('taxi-distance.csv', header=None).squeeze()
s1
```

```python
s2 = pd.read_csv('taxi-passenger-count.csv', header=None).squeeze()
s2
```

```python
df = pd.concat([s1, s2], axis=1)
df
```

```python
df.columns = ['distance', 'passengers_cnt']
```

```python
df
```

```python
iqr = df['distance'].quantile(0.75) - df['distance'].quantile(0.25)
```

```python
iqr
```

```python
outliers = df.loc[(df['distance'] < df['distance'].quantile(0.25) - 1.5 * iqr) | (df['distance'] > df['distance'].quantile(0.75) + 1.5 * iqr)]
```

```python
len(outliers)
```
