```python
def get_mean_comp(df, comb):
    lang1, lang2 = comb
    return df.loc[((df['LanguageHaveWorkedWith'].str.split(';').str.len() == 2) & (df[lang1] == 1) & (df[lang2] == 1))]['CompTotal'].mean()
```

```python
df2 = pd.read_csv(
    'data/so_2021_survey_results.csv',
    usecols=['LanguageHaveWorkedWith', 'LanguageWantToWorkWith', 'Country', 'CompTotal', 'NEWStuck', 'Gender', 'YearsCode', 'YearsCodePro']
)

df2
```

```python
df2['YearsCodePro'].value_counts()
```

```python
df2.loc[df2['YearsCode'] == 'Less than 1 year', 'YearsCode'] = 0
df2.loc[df2['YearsCode'] == 'More than 50 years', 'YearsCode'] = 51

df2.loc[df2['YearsCodePro'] == 'Less than 1 year', 'YearsCodePro'] = 0
df2.loc[df2['YearsCodePro'] == 'More than 50 years', 'YearsCodePro'] = 51
```

```python
df2 = df2.dropna(subset=['YearsCodePro', 'YearsCode'])
```

```python
df2.loc[:, 'YearsCodePro'] = df2['YearsCodePro'].astype(int)
df2.loc[:, 'YearsCode'] = df2['YearsCode'].astype(int)
```

```python
df2.loc[:, 'years_share'] = df2['YearsCodePro'] / df2['YearsCode']
```

```python
df2.loc[df2['YearsCode'] != 0, 'years_share'].mean()
```
