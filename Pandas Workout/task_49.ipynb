```python
import numpy as np
import pandas as pd
```

```python
import time
```

```python
start = time.perf_counter()
df = pd.read_csv(
    'data/nyc-parking-violations-2020.csv',
    low_memory=False
)
end = time.perf_counter()
print(f'{end - start}')
```

```python
start = time.perf_counter()
df.to_csv('csv_tmp.csv')
end = time.perf_counter()
print(f'{end - start}')
```

```python
start = time.perf_counter()
df.to_json('json_tmp.json')
end = time.perf_counter()
print(f'{end - start}')
```

```python
start = time.perf_counter()
df.to_feather('feather_tmp.feather')
end = time.perf_counter()
print(f'{end - start}')
```
