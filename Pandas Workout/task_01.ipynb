```python
import numpy as np
import pandas as pd
```

```python
g = np.random.default_rng(0)
```

```python
months = ['September', 'October', 'November', 'December', 'January', 'February', 'March', 'April', 'May', 'June']
```

```python
len(months)
```

```python
scores = g.integers(70, 101, 10)
```

```python
scores
```

```python
s = pd.Series(data=scores, index=months, dtype=np.int8)
s
```
