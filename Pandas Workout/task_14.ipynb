```python
import numpy as np
import pandas as pd
```

```python
s = pd.read_csv('data/nyc-temps.txt').squeeze()
s
```

```python
idx = list(range(0, 22, 3))
idx
```

```python
idx *= 91
```

```python
s.name = 'temp'
```

```python
s1 = pd.Series(data=idx, name='hour')
s1
```

```python
df = pd.concat([s, s1], axis=1)
df
```

```python
df.loc[df['temp'] < 0, 'temp'] = 0
```

```python
df
```

```python
df['temp'].value_counts()
```
