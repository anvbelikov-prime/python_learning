```python
df.groupby('passenger_count')['total_amount'].mean().sort_index(ascending=False)
```

```python
df1 = df.sort_values(by=['passenger_count'], ascending=[True])
```

```python
df1
```

```python
def get_distance_category(dist):
    if 0 <= dist < 2:
        return 'short'
    elif 2 <= dist <= 10:
        return 'medium'
    else:
        return 'long'
```

```python
df1['trip_distance_group'] = df['trip_distance'].apply(get_distance_category)
df1
```

```python
df1.groupby('trip_distance_group')['passenger_count'].mean().sort_values(ascending=False)
```

```python
df1 = pd.read_csv('data/nyc_taxi_2019-01.csv', usecols=['passenger_count', 'trip_distance', 'total_amount'])
df1['year'] = 2019
```

```python
df = pd.concat(objs=[df1, df2], ignore_index=True)
df
```

```python
df.groupby('year')['total_amount'].mean()
```
