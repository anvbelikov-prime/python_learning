#### Z-тест

```python
def z_crit(alpha=0.05, two_sided=True):
    """Critical z for CI."""
    if two_sided:
        return stats.norm.ppf(1 - alpha / 2)
    return stats.norm.ppf(1 - alpha)

def p_value_from_z(z, alternative="two-sided"):
    """
    alternative: 'two-sided' | 'greater' | 'less'
    """
    z = float(z)
    if alternative == "two-sided":
        return 2 * (1 - stats.norm.cdf(abs(z)))
    if alternative == "greater":
        return 1 - stats.norm.cdf(z)
    if alternative == "less":
        return stats.norm.cdf(z)
    raise ValueError("alternative must be: 'two-sided', 'greater', 'less'")
```

```python
def ztest_mean_1samp(x, mu0, sigma, alpha=0.05, alternative="two-sided"):
    """
    One-sample z-test for mean with known population std (sigma).
    H0: mu = mu0
    """
    n = x.size
    xbar = x.mean()
    se = sigma / np.sqrt(n)
    z = (xbar - mu0) / se
    p = p_value_from_z(z, alternative=alternative)

    # CI for mu (always two-sided by convention)
    zc = z_crit(alpha=alpha, two_sided=True)
    ci = (xbar - zc * se, xbar + zc * se)

    return {
        "n": n, "xbar": xbar, "sigma": sigma, "se": se,
        "z": z, "p_value": p, "ci_mean": ci
    }
```

```python
x = rng.normal(0, 1, 100)
```

```python
res = ztest_mean_1samp(x, 1, 1)
print(res)
```

```python
x = rng.normal(0, 1, 10000)
print(x.mean())
res = ztest_mean_1samp(x, 0, 1)
print(res['p_value'])
```

```python
def ztest_mean_2samp(x1, x2, sigma1, sigma2, alpha=0.05, alternative="two-sided"):
    """
    Two-sample z-test for difference of means with known sigmas.
    H0: mu1 - mu2 = 0
    """
    n1, n2 = x1.size, x2.size
    m1, m2 = x1.mean(), x2.mean()
    diff = m1 - m2
    se = np.sqrt((sigma1**2)/n1 + (sigma2**2)/n2)
    z = diff / se
    p = p_value_from_z(z, alternative=alternative)

    zc = z_crit(alpha=alpha, two_sided=True)
    ci = (diff - zc * se, diff + zc * se)

    return {
        "n1": n1, "n2": n2, "mean1": m1, "mean2": m2,
        "diff": diff, "sigma1": sigma1, "sigma2": sigma2, "se": se,
        "z": z, "p_value": p, "ci_diff": ci
    }
```

```python
x1 = rng.normal(10, 15, 100)
x2 = rng.normal(12, 20, 100)
res = ztest_mean_2samp(x1, x2, 15, 20)
print(f'{res['p_value']:.10f}')
```
