```python
employees['Salary'] = employees['Salary'].fillna(0).astype(np.int64)
```

```python
employees.info()
```

```python
employees.nunique()
```

```python
employees['Gender'] = employees['Gender'].astype("category")
```

```python
employees.info()
```

```python
employees['Team'] = employees['Team'].astype("category")
```

```python
employees.info()
```

```python
employees
```

```python
employees[((employees['Gender'] == 'Female') & (employees['Team'] == 'Business Dev')) | ~(employees['Mgmt'])].head()
```

```python
team_lst = ['Legal', 'Marketing', 'Sales']
```

```python
employees[employees['Team'].isin(team_lst)]
```
