```python
df_json = pd.read_json('tv_shows.json')
df_json['shows'].loc[0]
```

```python
# 1
serials = pd.json_normalize(data=df_json['shows'], meta=['show', 'runtime', 'network'], record_path='episodes')
serials
```

```python
# 2
lst = []
for s in serials['show'].unique():
    lst.append(serials[serials['show'] == s])
```

```python
lst
```

```python
len(lst)
```

```python
excel_file = pd.ExcelWriter('My_Serials.xlsx')
excel_file
```

```python
for df in lst:
    name = df['show'].unique()[0]
    df.to_excel(excel_writer=excel_file, sheet_name=name, index=False)
```

```python
excel_file.close()
```
