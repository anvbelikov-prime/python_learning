```python
def single_or_multi(pokemon_type):
    if '/' in pokemon_type:
        return 'Multi'
    else:
        return 'Single'
```

```python
pokemon.apply(single_or_multi)
```

```python
(pokemon.apply(single_or_multi).value_counts(normalize=True) * 100).round(1)
```

```python
def count_types(pokemon_type):
    c_dict = Counter(pokemon_type)
    return c_dict['/']
```

```python
pokemon.dropna().apply(count_types).value_counts()
```

```python
def split_types(pokemon_type):
    return pokemon_type.split(' / ')
```

```python
pokemon.dropna().apply(split_types).explode().value_counts()
```
