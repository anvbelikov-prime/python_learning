```python
inspections = pd.read_csv('chicago_food_inspections.csv')
inspections
```

```python
inspections['Name']
```

```python
inspections['Name'] = inspections['Name'].str.strip()
inspections
```

```python
inspections['Name']
```

```python
for col in inspections.columns:
    inspections[col] = inspections[col].str.strip()
```

```python
inspections
```

```python
inspections['Name'] = inspections['Name'].str.lower().str.title()
```

```python
inspections
```

```python
inspections['Risk'].unique()
```

```python
inspections['Risk'].value_counts()
```

```python
inspections = inspections.dropna(subset=['Risk'])
```

```python
inspections
```

```python
inspections['Risk'].unique()
```

```python
inspections['Risk'] = inspections['Risk'].replace('All', 'Risk 4 (Extreme)')
```

```python
inspections['Risk'].unique()
```

```python
inspections['Risk Num'] = inspections['Risk'].str.slice(5, 6)
```

```python
inspections['Risk Category'] = inspections['Risk'].str.slice(8, -1)
```
