```python
customers_both_weeks_and_dishes = weeks.loc['Week 1'].merge(weeks.loc['Week 2'], how='inner', on=['Customer ID', 'Food ID'])['Customer ID'].unique()
customers_both_weeks_and_dishes
```

```python
customers_one_week_only = weeks.loc[('Week 1',)].merge(weeks.loc[('Week 2',)], how='outer', on=['Customer ID'], indicator=True)
customers_one_week_only = customers_one_week_only[customers_one_week_only['_merge'].isin(['left_only', 'right_only'])]['Customer ID'].unique()
customers_one_week_only
```
