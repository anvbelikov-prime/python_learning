```python
df = pd.read_csv('citibike.csv', parse_dates=['start_time', 'stop_time'])
df
```

```python
df['start_day_name'] = df['start_time'].dt.day_name()
```

```python
df.pivot_table(index=['start_day_name'], values=['start_time'], aggfunc='count').sort_values(by=['start_time'], ascending=False)
```

```python
df.groupby(by=['start_day_name']).size().sort_values(ascending=False)
```

```python
df['start_day_name'].value_counts()
```
