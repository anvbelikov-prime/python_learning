```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('sales_by_employee.csv', parse_dates=['Date'])
df
```

```python
df.head()
```

```python
df.pivot_table(index=['Date'], values=['Revenue', 'Expenses'], aggfunc='mean')
```

```python
df.pivot_table(index='Date', values=['Revenue', 'Expenses'], aggfunc='mean')
```

```python
df.pivot_table(index=['Date'], values=['Revenue', 'Expenses'], aggfunc='sum')
```

```python
df.pivot_table(
    index=['Date'],
    columns=['Name'],
    values=['Revenue'],
    aggfunc='sum',
    fill_value=0.0,
    margins=True,
    margins_name='Total'
)
```

```python
df.pivot_table(
    index=['Date'],
    columns=['Name'],
    values=['Revenue'],
    aggfunc=['sum', 'count', 'mean'],
    fill_value=0.0,
    margins=True,
    margins_name='Total'
)
```
